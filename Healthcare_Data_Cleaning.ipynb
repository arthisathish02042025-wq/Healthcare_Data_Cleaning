```python
#Creating dataframe for Healthcare clinic

import pandas as pd
import numpy as np

data = {
    'patient_id': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
                   111, 112, 113, 114, 115, 101, 107, 118, 119, 120],
    'age': ['25', '34', None, '45', '29', None, '38', '52', '27', '41',
            '33', 'unknown', '48', '26', '35', '25', '38', '31', None, '44'],
    'weight': ['70', '65', '80', None, '75', None, '68', '90', '72', '85',
               '78', None, '82', '69', 'N/A', '70', '68', '74', None, '88'],
    'blood_pressure': [120, 130, None, 140, 125, None, 135, None, 118, 145,
                      128, None, 138, 122, None, 120, 135, 126, None, 142],
    'medication': ['Aspirin', 'Metformin', 'Lisinopril', None, 'Aspirin',
                   'Metformin', 'Lisinopril', 'Aspirin', None, 'Metformin',
                   'Lisinopril', 'Aspirin', None, 'Metformin', 'Aspirin',
                   'Aspirin', 'Lisinopril', 'Metformin', 'Aspirin', None],
    'insurance_provider': ['Blue Cross', 'Aetna', 'Cigna', 'UnitedHealth', None,
                          'Blue Cross', 'Aetna', 'Cigna', 'UnitedHealth', 'Blue Cross',
                          'Aetna', None, 'UnitedHealth', 'Blue Cross', 'Aetna',
                          'Blue Cross', 'Aetna', 'Cigna', 'UnitedHealth', None]
}

df = pd.DataFrame(data)
```

Task 1: Inspect the Data

```python
print("--- Task 1: Data Inspection ---")
print(df.info())
print("\nMissing values per column:\n", df.isnull().sum())
print("\nPercentage of missing values:\n", (df.isnull().sum() / len(df)) * 100)
print("\nNumber of duplicate rows:", df.duplicated().sum())
```

Task 2: Data Type Conversion

```python
#Before conversion displaying the table content
print(df)
```

```python
# --- Task 2: Data Type Conversion ---
print("\n--- Task 2: Data Type Conversion ---")
# Tracking existing missing values to count "new" ones created by conversion
age_nulls_before = df['age'].isnull().sum()
weight_nulls_before = df['weight'].isnull().sum()

# Convert age and weight to numeric (errors='coerce' turns strings like 'unknown' to NaN)
df['age'] = pd.to_numeric(df['age'], errors='coerce')
df['weight'] = pd.to_numeric(df['weight'], errors='coerce')

#Print New missing Value created in age and weight column
print(f"New missing values created in 'age': {df['age'].isnull().sum() - age_nulls_before}")
print(f"New missing values created in 'weight': {df['weight'].isnull().sum() - weight_nulls_before}")

# Convert insurance_provider datatype from object to category
df['insurance_provider'] = df['insurance_provider'].astype('category')

#Verify all concersions
print("\nVerified Data Types:\n", df.dtypes)
```

```python
#After conversion displaying the table content
print(df)
```

Task 3: Handle Missing Values

```python
# --- Task 3: Handle Missing Values ---
print("\n--- Task 3: Handling Missing Values ---")
# Imputing numeric values with median
df['age'] = df['age'].fillna(df['age'].median())
df['weight'] = df['weight'].fillna(df['weight'].median())
df['blood_pressure'] = df['blood_pressure'].fillna(df['blood_pressure'].median())

# Imputing categorical values (Fill with mode)
df['medication'] = df['medication'].fillna(df['medication'].mode()[0])

# First, add 'Unknown' to the list of allowed categories(as insurance_provider datatype is change from onject to category)
df['insurance_provider'] = df['insurance_provider'].cat.add_categories('Unknown')

# Now you can safely fill the missing values
df['insurance_provider'] = df['insurance_provider'].fillna('Unknown')
```

```python
#Displaying df after Handling Missing Values
print(df)
```

Task 4: Handle Duplicates

```python
#--- Task 4: Handle Duplicates ---
print("\n--- Task 4: Handle Duplicates ---")
# 1. View all duplicate rows (shows both the original and the copies)
print("All duplicate rows in the dataset:")
print(df[df.duplicated(keep=False)])

# 2. Identify duplicates based ONLY on patient_id
# This finds cases where the same ID appears twice, even if other data differs
print("\nDuplicates based on patient_id only:")
print(df[df.duplicated(subset=['patient_id'], keep=False)])

# 3. Remove duplicates and show the impact on data size
shape_before = df.shape
df = df.drop_duplicates(subset=['patient_id'], keep='first')
shape_after = df.shape

print(f"\nRows before: {shape_before[0]}")
print(f"Rows after: {shape_after[0]}")
print(f"Total rows removed: {shape_before[0] - shape_after[0]}")
```

```python
#Displaying df after Handling Duplicates
print(df)
```

```python
import pandas as pd
import numpy as np

# --- 1. Start Fresh: Load Data and Create Copy ---
data = {
    'patient_id': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112, 113, 114, 115, 101, 107, 118, 119, 120],
    'age': ['25', '34', None, '45', '29', None, '38', '52', '27', '41', '33', 'unknown', '48', '26', '35', '25', '38', '31', None, '44'],
    'weight': ['70', '65', '80', None, '75', None, '68', '90', '72', '85', '78', None, '82', '69', 'N/A', '70', '68', '74', None, '88'],
    'blood_pressure': [120, 130, None, 140, 125, None, 135, None, 118, 145, 128, None, 138, 122, None, 120, 135, 126, None, 142],
    'medication': ['Aspirin', 'Metformin', 'Lisinopril', None, 'Aspirin', 'Metformin', 'Lisinopril', 'Aspirin', None, 'Metformin', 'Lisinopril', 'Aspirin', None, 'Metformin', 'Aspirin', 'Aspirin', 'Lisinopril', 'Metformin', 'Aspirin', None],
    'insurance_provider': ['Blue Cross', 'Aetna', 'Cigna', 'UnitedHealth', None, 'Blue Cross', 'Aetna', 'Cigna', 'UnitedHealth', 'Blue Cross', 'Aetna', None, 'UnitedHealth', 'Blue Cross', 'Aetna', 'Blue Cross', 'Aetna', 'Cigna', 'UnitedHealth', None]
}

df = pd.DataFrame(data)
df_clean = df.copy()

# Capture "Before" metrics
shape_before = df.shape
missing_before = df.isnull().sum().sum()
duplicates_before = df.duplicated().sum()
dtypes_before = df.dtypes.copy()

# --- 2. Perform ALL Cleaning Steps ---

# Step A: Data Type Conversion
df_clean['age'] = pd.to_numeric(df_clean['age'], errors='coerce')
df_clean['weight'] = pd.to_numeric(df_clean['weight'], errors='coerce')
df_clean['insurance_provider'] = df_clean['insurance_provider'].astype('category')

# Step B: Handle Missing Values
# Fill numeric with median
df_clean['age'] = df_clean['age'].fillna(df_clean['age'].median())
df_clean['weight'] = df_clean['weight'].fillna(df_clean['weight'].median())
df_clean['blood_pressure'] = df_clean['blood_pressure'].fillna(df_clean['blood_pressure'].median())

# Fill medication with mode
df_clean['medication'] = df_clean['medication'].fillna(df_clean['medication'].mode()[0])

# Fill insurance_provider (Categorical Fix)
df_clean['insurance_provider'] = df_clean['insurance_provider'].cat.add_categories('Unknown')
df_clean['insurance_provider'] = df_clean['insurance_provider'].fillna('Unknown')

# Step C: Remove Duplicates (based on patient_id)
df_clean = df_clean.drop_duplicates(subset=['patient_id'], keep='first')

# --- 3. Verification Report ---
print("=== DATA CLEANING VERIFICATION REPORT ===")
print(f"Shape:            {shape_before} -> {df_clean.shape}")
print(f"Total Missing:    {missing_before} -> {df_clean.isnull().sum().sum()}")
print(f"Total Duplicates: {duplicates_before} -> {df_clean.duplicated().sum()}")
print("\n--- Data Types Comparison ---")
comparison = pd.DataFrame({'Before': dtypes_before, 'After': df_clean.dtypes})
print(comparison)
```
